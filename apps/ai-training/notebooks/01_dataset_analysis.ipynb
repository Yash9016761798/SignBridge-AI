## SignBridge AI Training

Analyze the Exploration-Lab/iSign dataset for Indian Sign Language gesture recognition.

**Objectives:**
- Load dataset from HuggingFace or local source
- Analyze class distribution
- Check video quality and frame counts
- Identify data integrity issues
- Generate analysis report

```python
import sys
sys.path.insert(0, '../scripts')

from utils import load_config, setup_logging
from analyze_dataset import analyze_dataset_structure, generate_analysis_report

setup_logging(level='INFO')
config = load_config('dataset')
print(f'Dataset source: {config["source"]["type"]}')
```

```python
# Analyze dataset structure
source_type = config['source']['type']
results = analyze_dataset_structure(source_type, config)

# Generate report
report = generate_analysis_report(results)
print(report)
```

```python
# Visualize class distribution (if data available)
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

if results['class_distribution']:
    df = pd.DataFrame.from_dict(results['class_distribution'], orient='index', columns=['count'])
    df = df.sort_values('count', ascending=False)
    
    plt.figure(figsize=(14, 6))
    sns.barplot(x=df.index, y=df['count'])
    plt.xticks(rotation=90)
    plt.title('Class Distribution')
    plt.tight_layout()
    plt.show()
else:
    print('No class distribution data available')
```
