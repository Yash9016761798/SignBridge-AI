```python
import os
import subprocess
from pathlib import Path

DATASET_DIR = '/content/drive/MyDrive/SignBridgeAI/dataset'
VIDEO_OUTPUT = '/content/drive/MyDrive/SignBridgeAI/dataset/videos'
POSE_OUTPUT = '/content/drive/MyDrive/SignBridgeAI/dataset/poses'

os.makedirs(VIDEO_OUTPUT, exist_ok=True)
os.makedirs(POSE_OUTPUT, exist_ok=True)

def extract_rar(archive_path, output_dir):
    cmd = ['unrar', 'x', '-o+', archive_path, output_dir + '/']
    result = subprocess.run(cmd, capture_output=True, text=True, timeout=3600)
    return result.returncode == 0
```

```python
# Extract video archives
print('Extracting video archives...')
video_dir = os.path.join(DATASET_DIR, 'data', 'video')
if os.path.exists(video_dir):
    archives = sorted(Path(video_dir).glob('*.rar'))
    for i, archive in enumerate(archives):
        print(f'  [{i+1}/{len(archives)}] {archive.name}')
        success = extract_rar(str(archive), VIDEO_OUTPUT)
        print(f'    {"OK" if success else "FAILED"}')
else:
    print(f'  Video archives not found at {video_dir}')
```

```python
# Extract pose archives
print('Extracting pose archives...')
pose_dir = os.path.join(DATASET_DIR, 'data', 'pose')
if os.path.exists(pose_dir):
    archives = sorted(Path(pose_dir).glob('*.rar'))
    for i, archive in enumerate(archives):
        print(f'  [{i+1}/{len(archives)}] {archive.name}')
        success = extract_rar(str(archive), POSE_OUTPUT)
        print(f'    {"OK" if success else "FAILED"}')
else:
    print(f'  Pose archives not found at {pose_dir}')
```

```python
# Verify extraction
print('Extraction Results:')

video_count = len(list(Path(VIDEO_OUTPUT).rglob('*.mp4')))
pose_count = len(list(Path(POSE_OUTPUT).rglob('*.npy')))

print(f'  Videos: {video_count} files')
print(f'  Poses:  {pose_count} files')

# Storage
total = 0
for d in [VIDEO_OUTPUT, POSE_OUTPUT]:
    for f in Path(d).rglob('*'):
        if f.is_file():
            total += f.stat().st_size
print(f'  Storage: {total/1024**3:.2f} GB')
```
