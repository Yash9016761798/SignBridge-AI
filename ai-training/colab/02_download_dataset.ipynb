```python
# Download using huggingface_hub directly
from huggingface_hub import hf_hub_download, snapshot_download
import os

REPO_ID = 'Exploration-Lab/iSign'
OUTPUT_DIR = '/content/drive/MyDrive/SignBridgeAI/dataset'
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f'Downloading from: {REPO_ID}')
print(f'Output: {OUTPUT_DIR}')
```

```python
# Download CSV
print('[1/3] Downloading CSV...')
csv_path = hf_hub_download(
    repo_id=REPO_ID,
    filename='iSign_v1.1.csv',
    local_dir=OUTPUT_DIR,
    token=os.environ.get('HF_TOKEN'),
)
print(f'  CSV downloaded: {csv_path}')
```

```python
# Download video archives
print('[2/3] Downloading video archives...')
video_archives = [
    'data/video/video.mp4.part1.rar',
    'data/video/video.mp4.part2.rar',
    'data/video/video.mp4.part3.rar',
    'data/video/video.mp4.part4.rar',
    'data/video/video.mp4.part5.rar',
    'data/video/video.mp4.part6.rar',
]

for i, archive in enumerate(video_archives):
    print(f'  [{i+1}/{len(video_archives)}] {archive}')
    try:
        hf_hub_download(
            repo_id=REPO_ID,
            filename=archive,
            local_dir=OUTPUT_DIR,
            token=os.environ.get('HF_TOKEN'),
        )
        print(f'    OK')
    except Exception as e:
        print(f'    FAILED: {e}')
```

```python
# Download pose archives
print('[3/3] Downloading pose archives...')
pose_archives = [
    f'data/pose/pose.mp4.part{i}.rar' for i in range(1, 13)
]

for i, archive in enumerate(pose_archives):
    print(f'  [{i+1}/{len(pose_archives)}] {archive}')
    try:
        hf_hub_download(
            repo_id=REPO_ID,
            filename=archive,
            local_dir=OUTPUT_DIR,
            token=os.environ.get('HF_TOKEN'),
        )
        print(f'    OK')
    except Exception as e:
        print(f'    FAILED: {e}')
```

```python
# Verify downloads
import os
from pathlib import Path

print('Downloaded files:')
total_size = 0
for f in sorted(Path(OUTPUT_DIR).rglob('*')):
    if f.is_file():
        size = f.stat().st_size
        total_size += size
        print(f'  {f.relative_to(OUTPUT_DIR)}: {size/1024**2:.1f} MB')

print(f'\nTotal: {total_size/1024**3:.2f} GB')
```
